```python
import pandas as pd
bwd = 10

feat = 'return'
sig = (
    unv.bar
    .mutate(feat=feat)
    .mutate(feat=_['feat'].mean().over(ibis.trailing_window(preceding=(bwd-1), group_by=_.secid, order_by=_.close_time)))
    .filter(_['universe_mask'])
    .filter(_['close_time'] <= pd.Timestamp('2022-01-01'))
    .group_by('close_time')
    .mutate(
        rank=_['feat'].rank(),
    )
    .group_by('close_time')
    .mutate(
        zscore=(_.rank - _.rank.mean()) / _.rank.std()
    )
    .group_by('close_time')
    .mutate(
        wgt=_.zscore / _.zscore.abs().sum()
    )
    .group_by('close_time')
    .mutate(
        # wgt=_.wgt * -1
        wgt=_.wgt
    )
    # .group_by('secid')
    .order_by('close_time', 'wgt')
)

bt = unv.backtest(sig, fwd=1)

sig
bt.bt_dly_perf_stats
bt.bt_dly_df.cumsum().plot()
```

```python
bwd = 240
fwd = 60

bwd = 1
fwd = 1

# fwd = 180

# # GOOD NET SHARPE
# bwd = 30
# fwd = 30

# bwd = 5
# fwd = 5

apply_universe = True
# start_date = '2020-02-01'
# end_date = '2020-12-31'
# end_date = '2020-03-31'
# end_date = '2020-06-30'

tcost_model = .0002
tcost_model = .0006

feat = 'residual'
# feat = 'return'
# 
sig = (
    unvm
    .bar
    .mutate(feat=_[feat])
    .mutate(feat=_.feat.mean().over(ibis.trailing_window(preceding=(unvm.risk_calculation_window-1), group_by=_.secid, order_by=_.close_time)))
    .mutate(feat=_.feat / _.residual_volatility)
    # .filter(_.unv_mask if apply_universe else True)
    .select(np.unique(['secid', 'open_time', 'close_time', 'close', 'return', 'universe_weight', 'universe_mask', 'feat']))
    .group_by('close_time')
    .mutate(
        rank=_.feat.rank(),
    )
    .group_by('close_time')
    .mutate(
        zscore=(_.rank - _.rank.mean()) / _.rank.std()
    )
    .group_by('close_time')
    .mutate(
        wgt=_.zscore / _.zscore.abs().sum()
    )
    .group_by('close_time')
    .mutate(
        wgt=_.wgt * -1
        # wgt=_.wgt
    )
    .group_by('secid')
    .order_by('close_time')
    .mutate(
        # feat=_.pct_change.sum().over(ibis.trailing_window(preceding=(bwd-1), group_by=_.secid, order_by=_.close_time)),
        wgt_agg=_.wgt.mean().over(ibis.trailing_window(preceding=(fwd-1), group_by=_.secid, order_by=_.close_time)),
    )
    .group_by('secid')
    .order_by('close_time')
    .mutate(
        pnl_wgt_agg=_.wgt_agg.lag()
    )
    # .mutate(
    #     # fwd_ret_1p=_.pct_change.lag(-1),
    #     # fwd_ret_5p=_.pct_change_5p.lag(-5)
    #     **{f'fwd_ret_{fwd}p': _[f'pct_change_{fwd}p'].lag(-fwd)}
    # )
    # .group_by('secid')
    # .order_by('close_time')
    # .mutate(fwd_ret_1p=_.pct_change.lag(-1))
)

pnl = (
    sig
    .group_by('secid')
    .order_by('close_time')
    .mutate(
        pnl_gross=_.pnl_wgt_agg * _['return'],
        turnover=_.pnl_wgt_agg - _.pnl_wgt_agg.lag(),
        tcost_model=tcost_model
    )
    .mutate(
        tcost=_.turnover.abs() * _.tcost_model
    )
    .mutate(
        pnl_net=_.pnl_gross - _.tcost
    )
)

bt = (
    pnl
    .group_by('close_time')
    .aggregate(
        por_ret_gross=_.pnl_gross.sum(),
        por_ret_net=_.pnl_net.sum(),
    )
    .order_by('close_time')
)

bt_dly = (
    bt
    .group_by(_.close_time.date(), )
    .aggregate(
        por_ret_gross=_.por_ret_gross.sum(),
        por_ret_net=_.por_ret_net.sum(),
    )
    .mutate(
        close_time='Date(close_time)'
    )
    .order_by('close_time')
)

bt_df = bt.select(['close_time', 'por_ret_gross', 'por_ret_net']).to_pandas().set_index('close_time')
bt_dly_df = bt_dly.select(['close_time', 'por_ret_gross', 'por_ret_net']).to_pandas().set_index('close_time')
```
